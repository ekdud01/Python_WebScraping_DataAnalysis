### 1-2. Daum 뉴스 기사 제목 스크래핑하기 코드를 섹션 별로 처리하는 함수로 구현하기

```python
# 경제 뉴스 URL
url = 'https://news.daum.net/economy'
# 사회 뉴스 URL
url = 'https://news.daum.net/society'
```

```python
import requests
from bs4 import BeautifulSoup

section_dict = {'기후/환경':'climate', '사회':'society', '경제':'economy', '정치':'politics',\
                '국제':'world', '문화':'culture', '생활':'life', 'IT/과학':'tech', '인물':'people'}

def print_news(section_name):
    sid = section_dict.get(section_name, 'society')
    url = f'https://news.daum.net/{sid}'
    print(f'=====> {url} {section_name} 뉴스 <=====')

    # 요청 헤더 설정 : 브라우저 정보
    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
    }

    # requests 의 get() 함수 호출하기 
    res = requests.get(url, headers=req_header)

    if not res.ok:
        print(f'Error Code = {res.status_code}')
        exit()

    soup = BeautifulSoup(res.content, 'html.parser')

    li_tags = soup.select("ul.list_newsheadline2 li")

    for li_tag in li_tags:
        a_tag = li_tag.find('a')
        link = a_tag['href']

        strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
        title = strong_tag.text

        print(link)
        print(title)

    print()
```

```python
print_news('경제')
print_news('사회')
```

# 2. 웹스크래핑 연습문제 2

### 2-1. Nate 뉴스 기사 제목 스크래핑하기 (선택)

#### : 6개의 섹션의 뉴스를 출력하는 함수를 생성하여 스크래핑하기
* 최신 뉴스, 정치, 경제, 사회, 세계, IT/과학

```python
# Nate 뉴스 URL
url = 'https://news.nate.com/recent?mid=n0100'
```

* 뉴스 기사의 Image를 출력하세요.
1. Image의 도메인 이름이 포함된 url과 src속성의 img 경로를 합치려면, urljoin 함수를 사용하세요.
2. Image 출력은 Image 클래스와 display 함수를 사용하세요.
3. img 엘리먼트의 존재 여부를 체크하신 후에 src 속성의 이미지를 경로를 추출하기

```python
from urllib.parse import urljoin
from IPython.display import Image, display
import requests
from bs4 import BeautifulSoup

section_dict = {'최신뉴스':'n0100', '정치':'n0200', '경제':'n0300', '사회':'n0400', '세계':'n0500', 'IT/과학':'n0600'}

def print_nate_news(section_name):
    sid = section_dict.get(section_name, 'n0100')

    if section_name == '최신뉴스':
        url = f'https://news.nate.com/recent?mid={sid}'
    else:
        url = f'https://news.nate.com/section?mid={sid}'
    print(f'====================> {section_name} 뉴스 {url} <====================')

    # 요청 헤더 설정 : 브라우저 정보
    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
    }

    # requests 의 get() 함수 호출하기 
    res = requests.get(url, headers=req_header)

    if not res.ok:
        print(f'Error Code = {res.status_code}')
        exit()

    soup = BeautifulSoup(res.content, 'html.parser')
    li_tags = soup.select('div.mlt01')

    for li_tag in li_tags:
        a_tag = li_tag.select_one('a.lit') or li_tag.select_one('a')

        title_tag = li_tag.select_one('h2.tit, .tit') 
        if title_tag:
            title = ' '.join(title_tag.text.split())
        else:
            title = ' '.join(a_tag.text.splitlines()[0].split())

        link = urljoin(url, a_tag['href'])

        print(link)
        print(title)

        img_tag = li_tag.select_one('img')

        if img_tag and img_tag.get('src'):
            img_src = img_tag['src']
            img_url = urljoin(url, img_src)

            display(Image(url=img_url))
            print(f'이미지 URL: {img_url}')
        else:
            print('[ 이미지 없음 ]')

        print("----------" * 10)
    print()
```

```python
print_nate_news('최신뉴스')
print_nate_news('정치')
print_nate_news('경제')
print_nate_news('사회')
print_nate_news('세계')
print_nate_news('IT/과학')
```

### 2-2. 하나의 네이버 웹툰과 1개의 회차에 대한 Image 다운로드하기 (필수)

#### : 하나의 웹툰의 제목(title)과 회차번호(no), 회차의 URL(url)을 입력으로 받는 함수를 선언합니다.
* def download_one_episode(title, no, url):

```python
import requests
from bs4 import BeautifulSoup
import os

def download_one_episode(title, no, url):
    req_header = {
        'referer': url,
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
    }

    # img\title(웹툰 제목)\no(회차) 디렉토리 경로
    imgdir_name = os.path.join('img', title, str(no))
    # 이미지 저장 폴더가 없으면 생성 (sub 디렉토리도 생성)
    os.makedirs(imgdir_name, exist_ok=True)

    # 웹 페이지 요청 및 확인
    res = requests.get(url)
    if not res.ok:
        print(f'Error Code = {res.status_code}')
        exit()

    # 이미지 URL 추출
    soup = BeautifulSoup(res.text, 'html.parser')
    img_tags = soup.select("img[src*='IMAG']")

    # 이미지 다운로드
    for idx, img_tag in enumerate(img_tags, start=1):
        img_url = img_tag.get('src')
        if not img_url:
            continue

        res = requests.get(img_url, headers=req_header)
        if res.ok:
            img_data = res.content

            # img\title(웹툰 제목)\no(회차)\xxx.jpg
            file_name = f"{idx:03d}.jpg"
            file_path = os.path.join(imgdir_name, file_name)
            
            with open(file_path, 'wb') as file:
                print(f'Writing to {file_path} ({len(img_data):,} bytes)')
                file.write(img_data)
        else:
            print(f'Error Code = {res.status_code} for {img_url}')
```

```python
download_one_episode('일렉시드', 341, 'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed')
```
