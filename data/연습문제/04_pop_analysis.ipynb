```python
import pandas as pd

data = pd.read_csv('../인구현황.csv')
```

```python
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
```

```python
# 폰트이름과 폰트파일 정보 출력 list comprehension 
[ (font.name, font.fname) for font in fm.fontManager.ttflist if 'Mal' in font.name ]
```

```python
# font name을 알고 있다면 생략가능
# 한글폰트 path 설정
font_path = 'C:\\Windows\\Fonts\\malgun.ttf'
# font의 파일 정보로 font name을 알아내기
font_name = fm.FontProperties(fname=font_path).get_name()
print(font_name)
```

```python
# matplotlib의 rc(run command) 함수를 사용해서 font name(Malgun Gothic) 설정
matplotlib.rc('font', family=font_name)
```

##### 1. 지역별 총 인구수 Top 5

```python
p_data = data[data['행정기관'] != '전국'].copy()
p_data.head()
```

```python
data.columns
```

```python
p_data[['행정기관', '총인구수']].sort_values(by='총인구수', ascending=False).head(5)
```

##### 1-1. 지역별 총 인구수 plot 

```python
g_df = p_data.loc[p_data['행정기관'] != '전국'].copy()
pop_order = g_df.sort_values('총인구수',ascending=False)['행정기관']

# subplots() 함수를 사용하여 axes 객체를 생성하기
figure, axes1 = plt.subplots(nrows=1,ncols=1)
figure.set_size_inches(18,12)

# seaborn의 barplot() 함수 호출하기 - 행정기관과 총인구수
sns.barplot(data=g_df, x="행정기관", y="총인구수", ax=axes1, hue="행정기관", order=pop_order)

axes1.set_title(f'지역별 총인구수')
axes1.set_xticklabels(axes1.get_xticklabels(), rotation=45, horizontalalignment='right', fontsize=12)

plt.show()
```

##### 2. 세대당 인구수 높은 지역 Top 5

```python
p_data[['행정기관', '세대당 인구']].sort_values(by='세대당 인구', ascending=False).head(5).reset_index(drop=True)
```

##### 2-1. 지역별 세대당 인구 plot

```python
area_order = g_df.sort_values('세대당 인구',ascending=False)['행정기관']

# subplots() 함수를 사용하여 axes 객체를 생성하기
figure, axes1 = plt.subplots(nrows=1,ncols=1)
figure.set_size_inches(18,12)

# seaborn의 barplot() 함수 호출하기 - 행정기관과 세대당 인구
sns.barplot(data=g_df, x="행정기관", y="세대당 인구", ax=axes1, hue="행정기관", order=area_order)

axes1.set_title(f'지역별 세대당 인구')
axes1.set_xticklabels(axes1.get_xticklabels(), rotation=45, horizontalalignment='right', fontsize=12)

plt.show()
```

##### 3. 남녀 비율 분석

```python
p_data[['행정기관', '남여 비율']].sort_values(by='남여 비율', ascending=False).reset_index(drop=True)
```

##### 3-1. 남녀 비율 분석 plot

```python
percent_order = g_df.sort_values('남여 비율',ascending=False)['행정기관']

# subplots() 함수를 사용하여 axes 객체를 생성하기
figure, axes1 = plt.subplots(nrows=1,ncols=1)
figure.set_size_inches(18,12)

# seaborn의 barplot() 함수 호출하기 - 행정기관과 남여 비율
sns.barplot(data=g_df, x="행정기관", y="남여 비율", ax=axes1, hue="행정기관", order=percent_order)

# pyplot의 axhline() 함수를 사용하여 기준점 생성하기 
axes1.axhline(1.0, color='red', linestyle='--', label='남녀 1:1 비율')
axes1.legend(loc='upper right', fontsize=12)

axes1.set_title(f'지역별 남녀 비율')
axes1.set_xticklabels(axes1.get_xticklabels(), rotation=45, horizontalalignment='right', fontsize=12)

plt.show()
```

##### 4. 남자초과 / 여자초과 분석

```python
import numpy as np

conditions = [p_data['남여 비율'] > 1, p_data['남여 비율'] < 1, p_data['남여 비율'] == 1]
choices = ['남초', '여초', '동일']

p_data['남초여초'] = np.select(conditions, choices)

p_data[['행정기관', '남여 비율', '남초여초']]
```

##### 5. 세대당 인구 평균보다 높은 지역

```python
p_data.loc[p_data['세대당 인구'] >= p_data['세대당 인구'].mean(), ['행정기관', '세대당 인구']]\
           .sort_values('세대당 인구', ascending=False).reset_index(drop=True)
```

##### 5-1. 남초 vs 여초 개수 plot (seaborn의 countplot) | 지역별 세대수 plot (seaborn의 barplot)

```python
# subplots() 함수를 사용하여 axes 객체를 생성하기
figure,(axes1,axes2) = plt.subplots(nrows=2,ncols=1)
figure.set_size_inches(18,12)

count_plot = sns.countplot(x='남초여초', data=p_data, order=['여초','동일','남초'], ax=axes1, hue='남초여초')
count_plot.set_title(f'남초 vs 여초 지역 개수')

pop_plot = sns.barplot(x='행정기관', y='세대수', data=g_df.sort_values(by='세대수',ascending=False), ax=axes2, hue='행정기관')
pop_plot.set_title(f'지역별 세대수')

plt.show()
```
